```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("cleaned_data.csv")
```

BAR CHART   

TOTAL NET AMOUNT BY CATEGORY

```python
category_sales=df.groupby("Product Category")["Net Amount"].sum()
plt.figure(figsize=(10,6))
plt.bar(category_sales.index,category_sales.values)
plt.title("Total Net Amount By Category")
plt.xlabel("product category")
plt.ylabel("total net amount")
plt.xticks(rotation=30)
plt.ticklabel_format(axis="y",style="plain")         #displays full numbers instead of scientific notations
plt.tight_layout()                                   #helps prevent your category labels from getting cut off
plt.show()
```

PIE CHART

PURCHASE METHOD DISTRIBUTION

```python
purchase_counts=df["Purchase Method"].value_counts()
plt.figure(figsize=(10,6))
plt.pie(
    purchase_counts.values,
    labels=purchase_counts.index,
    autopct="%1.1f%%"
)
plt.title("purchase method distribution")
plt.show()
```

SCATTER PLOT:-

CHECK RELATIONSHIP BETWEEN GROSS AMOUNT AND NET AMOUNT

```python
plt.Figure(figsize=(8,6))
plt.scatter(df["Gross Amount"],df["Net Amount"])
plt.title("gross amount vs net amount")
plt.xlabel("gross amount")
plt.ylabel("net amount")
plt.show()
```

BOX PLOT:-

NET AMOUNT DISTRIBUTION

```python
plt.figure(figsize=(8,6))
sns.boxplot(x= "Age Group",y="Net Amount",data = df)
plt.title("Net Amount Distribution by Age group")
plt.xlabel("Age Group")
plt.ylabel("Net Amount")
plt.xticks(rotation = 30)
plt.show()
```

The Net Amount is quite similar across all age groups. The median values and overall spread are nearly the same. There is one possible high-value outlier in the 45–60 age group. Overall age does not seem to make a major difference in transaction amounts.

HEATMAP:-

RELATIONSHIP BETWEEN GROSS AMOUNT,DISCOUNT AMOUNT,NET AMOUNT

```python
amounts=df[["Gross Amount","Discount Amount (INR)","Net Amount"]]
correlation=amounts.corr()
plt.figure(figsize=(8,6))
sns.heatmap(correlation,annot=True,cmap="coolwarm",vmin=-1,vmax=1)
plt.title("Correlation Between Transaction Amounts")
plt.tight_layout()
plt.show()
```
